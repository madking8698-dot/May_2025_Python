- When we open file three argumets are important 
- file: which needs file location
    - model:- 
    - 'r':- open for reading(default)
    - 'w':- open for writting, truncate  the file first 
    - 'x':- create a new file and  open it for writting 
    - 'a':- open for writting,appending to the end of the file if it exists 
    - 'b':- binary mode 
    - 't':- text mode(default)
    - '+':- open a disk file for updating(reading and writting)

- Encoding:- 
    - some student able to see spl chracter starting of the file 
    - this we can avoid using encoding 
    - the best one is : **utf-8-sig

```python
file_location=r'mobx_short_notes.txt'
file=open(file_location,
         mode='r',
         encoding='utf-8-sig')
data=file.read()
print(data)
```
