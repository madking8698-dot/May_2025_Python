**Note:**
- / beefore dont use argument name
- /after you can use argument name,that is our wish
-   '*' indicates no need to maintain the order of the arguments

```python
def add(a,/,*,b,c):
    print(a+b+c)

add(10,c=20,b=30)
```

```python
add(10,20,b=30)
```

```python
add(10,20)
```

```python
add(10,20,30)
```

**NOTE**
- (*) indicates provide the argument 

```python
sorted('python',reverse=True)

# / before dont use ==========> correct 
# * after you should use =====> wrong 
```

```python
sorted('python',reverse=True)
```

```python
sorted('python')                      # Works
sorted(iterable='python')             # Fail
sorted('python',reverse=True)         # Works
sorted(iterable='python',reverse=True)# fails
sorted('python',True)                 # Fails
# by default reverse= false 
```
