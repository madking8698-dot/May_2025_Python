**sliding window appraoch**
- fix the slice value first : **str1[i:i+3]**
- then increase the value by 1 **for i in range(len(str1))**
- whenever it is equql to substring **str1[i:i+3]=='ola'** 
- then update the count  **count=count+1**


```python
# str1[i:i+3]=='ola' logic
# i=0 str1[0:3]=='ola' true
# i=1 str1[1:3]= 'la' is not equal to'ola' false 

str1='ola ola ola'
count=0
for i in range(len(str1)):
    if str1[i:i+3]=='ola':
        count=count+1
print(count)
```
