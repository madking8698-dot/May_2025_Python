```python
# Q8) 
# ask user get 5 random numbers
# means you need a get a random number inside for loop
# the number should 5 times 
# now perform the even and odd operation
# 1) even count # 2) sum of even numbers # 3) odd count 
# 4) sum of odd number 
```

```python
# even_count=0
#
```

```python
import random
even_count=0
odd_count=0
even_sum=0
odd_sum=0
for i in range(5):
    num=random.randint(1,100)
    if num%2==0:
        print(f"{num} is an even")
        even_count=even_count+1
        even_sum=even_sum+num
    else:
        print(f"{num} is an odd")
        odd_count=odd_count+1
        odd_sum=odd_sum+num
print('Number of even are:-',even_count)
print('Number of odds are:-',odd_count)
print('Number of even are:-',even_sum)
print('Number of even are:-',odd_sum)
```

```python
# Q9) Game program 
# User enter one number 
# another number will generate randomly 
# if both numbers are equal then you won
# otherwise you lost
# i want to give three chances 
# for loop
# num=random
# num2=user keyboard 
# if num1==num2
#   print won
# else 
#    print fail 
```

```python
import random
for i in range(3):
    n1=random.randint(1,10)
    n2=eval(input('Enter a number'))
    if n1==n2:
        print('You won')
        break
    else:
        print('Better luck next time...')
```
