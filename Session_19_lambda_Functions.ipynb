**Lambda Functions**
- lambda functions represents functions in a single line 
- it is similer analogy to list comprehension
- to reduce the time complexity we will go for lambda functions 

$ pattern-1$:**Function with one argument**

```python
def mul(a):
    return(a*a)
ans=mul(10)
ans
```

```python
# Syntax 
# fuctions_name= lambda<arguments>: <return output> 
# in the above 
# fuction_name= mul
# argument name=a
# return output = a*a
```

```python
mul=lambda a: a*a
mul(20)
```

```python
def cube(a):
    return(a*a*a)
cube(100)
```

```python
cube=lambda a: a*a*a
cube(20)
```

$pattern- 2$: **function with two arguments**

```python
def add(a,b):
    return(a+b)
add(10,20)
```

```python
# syntax
# function name= lambda arg1,arg2:return_statement
```

```python
add=lambda a,b:a+b
add(10,100)
```

```python
average=lambda a,b,c:a+b+c/3
average(9,20,30)
```

```python
def avg(a,b,c):
    return(round((a+b+c)/3),2)
avg(9,23,21)
```

$pattern-:$ **Three arguments**

```python
avg=lambda a,b,c:(round((a+b+c)/3),2)
avg(199,232,21)
```

$pattern-4$: **Default arguments**

```python
avg=lambda a,b,c=21 : round((a+b+c)/3,2)
avg(199,232)
```

```python
mul=lambda a:a*a
add=lambda a,b : a+b
avg=lambda a,b,c : round((a+b+c)/3,2)
avg=lambda a,b,c=21 : round((a+b+c)/3,2)
```

**Filter**
- filter is apply on lambda function to retrive the True output
- map will just map the fuction and iterator 
- when we apply condition map will give True False 
- filter will use instead of map

```python
input=['hyd','beng#aluru','mumbai','pu#ne']
list(filter(lambda i: '#' in i,input))
```

```python
list(filter(lambda i: '#' not in i,input))
```
