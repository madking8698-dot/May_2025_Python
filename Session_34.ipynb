```python
# >85 A grade 60-85 B grade less than 60 c grde
```

```python
l=[45,55,65,75,85,95]
for i in l:
    if i>85:
        print(f'{i}:A Grade')
    elif i>60:
        print(f'{i}:B Grade')
    else:
        print(f'{i}:C Grade')
```

```python
# [<if out><if cond>else<elif cond> else <else out> <for loop >]
[f'{i}:A grade' if i>85 else f'{i}: B grade' if i>60 else f'{i}:C grade' for i in l]
```

**distance between two points**

```python
l1=[[1,2],[4,5],[7,9]]
l2=[10,20]
# find the distance between l2 with all other points from l1
# which is the maximum distance give me that 

```

```python
import math 
l1=[10,20]
l2=[30,40]
d1=l2[0]-l1[0] # (x2-x1)^2=30-10=20
d1_square=math.pow(d1,2)
d1_square
d2=l2[1]-l1[1] # (x2-x1)^2=30-10=20
d2_square=math.pow(d2,2)
d2_square
summ=d1_square+d2_square
math.sqrt(summ)
```
